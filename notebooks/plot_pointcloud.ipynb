```python
import open3d as o3d
import numpy as np
```

```python

# 1. Load the PLY file
print("Loading point cloud...")
pcd = o3d.io.read_point_cloud("pointcloud.ply")

# 2. Check if it loaded correctly (Open3D automatically ignores broken 'inf' points usually)
print(f"Loaded {len(pcd.points)} points.")

# 3. Visualization Window
# Mouse Controls: Left-Click (Rotate), Ctrl+Left (Pan), Wheel (Zoom)
o3d.visualization.draw_geometries([pcd], 
                                  window_name="PLY Preview",
                                  width=800,
                                  height=600)
```

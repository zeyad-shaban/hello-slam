```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/slam_trajectory_data.csv")

print(df.columns)
df.head()
```

```python
gt_x_arr, gt_y_arr, gt_z_arr = df['gt_tx'].to_numpy(), df['gt_ty'].to_numpy(), df['gt_tz'].to_numpy()
est_x_arr, est_y_arr, est_z_arr = df['est_tx'].to_numpy(), df['est_ty'].to_numpy(), df['est_tz'].to_numpy()
```

```python
fig = plt.figure(figsize=(6, 11))
ax = fig.add_subplot(projection='3d')

ax.plot(gt_x_arr, gt_y_arr, gt_z_arr, label="Grouth Truth", alpha=0.5, lw=2)
ax.plot(est_x_arr, est_y_arr, est_z_arr, label="Estimated Traj", lw=3, color="b")

ax.legend()
```
